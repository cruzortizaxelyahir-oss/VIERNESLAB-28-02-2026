# Actividad C. Limpieza y tratamiento de errores (25 puntos)

Crea `df_limpio` y realiza:
1. Limpieza de texto (`campus`, `modalidad`)
2. Normalización de etiquetas (ej. `slp centro`, `SLP Cntro`, `escolar`, `MIXTA`)
3. Eliminar duplicados por `id_estudiante`
4. Corregir rangos en `promedio` y `asistencia_pct` (0–100)
5. Corregir `creditos_aprobados <= creditos_inscritos`
6. Imputar faltantes (`edad`, `promedio`, `asistencia_pct`) con mediana

```python
# --- Respuesta del estudiante (C1) ---
# Crear df_limpio y realizar limpieza

# --- Respuesta del estudiante (C1) ---

# 1. Eliminar duplicados exactos
df_limpio = df.drop_duplicates().copy()

# 2. Estandarizar 'campus' y 'modalidad' (minúsculas y sin espacios extras)
df_limpio['campus'] = df_limpio['campus'].str.strip().str.lower()
df_limpio['modalidad'] = df_limpio['modalidad'].str.strip().str.lower()

# 3. Corregir inconsistencias específicas en 'campus' (ej. 'slp cntro' -> 'slp centro')
df_limpio['campus'] = df_limpio['campus'].replace({'slp cntro': 'slp centro'})

# 4. Imputación de valores nulos
# Edad con la mediana
df_limpio['edad'] = df_limpio['edad'].fillna(df_limpio['edad'].median())
# Promedio con la media
df_limpio['promedio'] = df_limpio['promedio'].fillna(df_limpio['promedio'].mean())
# Asistencia con 0 (asumiendo falta total si no hay registro)
df_limpio['asistencia_pct'] = df_limpio['asistencia_pct'].fillna(0)

# 5. Tratamiento de Outliers (Promedio y Asistencia no pueden exceder 100)
df_limpio['promedio'] = df_limpio['promedio'].clip(upper=100)
df_limpio['asistencia_pct'] = df_limpio['asistencia_pct'].clip(upper=100)

print("Limpieza completada. Registros finales:", df_limpio.shape[0])
```

# Actividad D. Filtración y transformación (20 puntos)

Con `df_limpio`:
1. Filtra `promedio < 80`
2. Filtra `campus == 'SLP Centro'` y `modalidad == 'Escolarizada'`
3. Crea `tasa_aprobacion = creditos_aprobados / creditos_inscritos`
4. Crea `riesgo_academico`:
   - Alto: `promedio < 70` o `asistencia_pct < 75`
   - Medio: `promedio < 80`
   - Bajo: otro caso
5. (Opcional) One-hot encoding de `modalidad`

```python
# --- Respuesta del estudiante (D1) ---
# Filtros solicitados
# --- Respuesta del estudiante (D1) ---

# 1. Crear columna 'eficiencia' (aprobados / inscritos)
df_limpio['eficiencia'] = df_limpio['creditos_aprobados'] / df_limpio['creditos_inscritos']

# 2. Crear columna 'estatus' mediante np.select
condiciones = [
    (df_limpio['promedio'] >= 90),
    (df_limpio['promedio'] >= 70) & (df_limpio['promedio'] < 90),
    (df_limpio['promedio'] < 70)
]
opciones = ['Excelente', 'Regular', 'Riesgo']
df_limpio['estatus'] = np.select(condiciones, opciones, default='Indefinido')

# 3. Filtrar estudiantes de 'slp centro' con promedio > 80
df_centro_top = df_limpio[(df_limpio['campus'] == 'slp centro') & (df_limpio['promedio'] > 80)]

display(df_centro_top)
```

```python
# --- Respuesta del estudiante (D2) ---
# Crear tasa_aprobacion y riesgo_academico

# Crear tasa_aprobacion
df_limpio["tasa_aprobacion"] = df_limpio["creditos_aprobados"] / df_limpio["creditos_inscritos"]

# Crear variable riesgo_academico
condiciones = [
    (df_limpio["promedio"] < 70) | (df_limpio["asistencia_pct"] < 75),
    (df_limpio["promedio"] < 80)
]

valores = ["Alto", "Medio"]

df_limpio["riesgo_academico"] = np.select(condiciones, valores, default="Bajo")
```

# Actividad E. Fusión (`merge`) y resumen final (15 puntos)

1. Une `df_limpio` con `apoyos` por `id_estudiante` con `how='left'` e `indicator=True`
2. Guarda en `df_final`
3. Muestra `shape` y conteo de `_merge`
4. Resume:
   - promedio general
   - promedio por `campus`
   - distribución de `riesgo_academico`
5. Guarda CSV en `salida_eval/estudiantes_preprocesados_eval.csv`

```python
# --- Respuesta del estudiante (E1) ---
# Merge y creación de df_final
# --- Respuesta del estudiante (E1) ---

# Ensure tasa_aprobacion and riesgo_academico are present in df_limpio
# This ensures df_final will have these columns even if D2 was not run or run out of order
df_limpio["tasa_aprobacion"] = df_limpio["creditos_aprobados"] / df_limpio["creditos_inscritos"]
condiciones_riesgo = [
    (df_limpio["promedio"] < 70) | (df_limpio["asistencia_pct"] < 75),
    (df_limpio["promedio"] < 80)
]
valores_riesgo = ["Alto", "Medio"]
df_limpio["riesgo_academico"] = np.select(condiciones_riesgo, valores_riesgo, default="Bajo")

# 1. Merge (Left Join) para mantener a todos los estudiantes de df_limpio
df_final = pd.merge(df_limpio, apoyos, on='id_estudiante', how='left')

# 2. Llenar nulos en 'beca' con 'No' y en 'tutoria_horas_mes' con 0
df_final['beca'] = df_final['beca'].fillna('No')
df_final['tutoria_horas_mes'] = df_final['tutoria_horas_mes'].fillna(0)

# 3. Resumen: Promedio de 'promedio' y 'asistencia_pct' por 'modalidad'
resumen_modalidad = df_final.groupby('modalidad')[['promedio', 'asistencia_pct']].mean()

print("Resumen por Modalidad:")
display(resumen_modalidad)

# Guardar el resultado final
df_final.to_csv("evaluacion_datos/estudiantes_final.csv", index=False)
print("Archivo 'estudiantes_final.csv' guardado ✅")
```

```python
# --- Respuesta del estudiante (E2) ---
# Resumen final

# Promedio general
print("Promedio general:")
print(df_final["promedio"].mean())

# Promedio por campus
print("\nPromedio por campus:")
print(df_final.groupby("campus")["promedio"].mean())

# Distribución de riesgo
print("\nDistribución de riesgo académico:")
print(df_final["riesgo_academico"].value_counts())
```

```python
# --- Respuesta del estudiante (E3) ---
# Guardar archivo final en CSV

from pathlib import Path

Path("salida_eval").mkdir(exist_ok=True)

df_final.to_csv(
    "salida_eval/estudiantes_preprocesados_eval.csv",
    index=False
)

print("Archivo guardado correctamente")
```
