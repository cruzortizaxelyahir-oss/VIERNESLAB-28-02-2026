```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Datos
cats = ['Norte', 'Sur', 'Este', 'Oeste']
conteos = np.array([120, 90, 150, 110])

# Gráfica
plt.figure(figsize=(8, 5))
plt.bar(cats, conteos, color=['#4e79a7', '#f28e2b', '#e15759', '#76b7b2'])
plt.title("Set 1: Distribución por Zonas Geográficas")
plt.ylabel("Conteos")
plt.show()
```

```python

# Datos
n2 = 120
x2 = np.linspace(0, 100, n2)
y2 = 2.0 * x2 + np.random.normal(0, 25, n2)

# Gráfica
plt.figure(figsize=(8, 5))
sns.regplot(x=x2, y=y2, scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.title("Set 2: Relación Lineal entre X y Y")
plt.show()
```

```python
# Datos
muestra = np.random.normal(loc=50, scale=10, size=1000) + np.random.exponential(scale=2, size=1000)*0.5

# Gráfica
plt.figure(figsize=(8, 5))
sns.histplot(muestra, kde=True, color="teal")
plt.title("Set 3: Histograma de Frecuencias y Densidad")
plt.show()
```

```python
# Datos
t4 = np.arange(1, 25)
serie = (5 + 1.2 * t4) + (4 * np.sin(2 * np.pi * t4 / 12)) + np.random.normal(0, 2, size=t4.shape)

# Gráfica
plt.figure(figsize=(10, 5))
plt.plot(t4, serie, marker='s', color='darkorange', linewidth=2)
plt.title("Set 4: Serie de Tiempo (Tendencia + Estacionalidad)")
plt.xlabel("Tiempo")
plt.show()
```

```python

# Datos
df = pd.DataFrame({
    'ingresos': np.random.normal(50, 10, 200),
    'gasto_marketing': np.random.normal(25, 5, 200),
    'clientes_nuevos': np.random.poisson(12, 200) + np.random.normal(0, 1, 200),
    'satisfaccion': np.clip(np.random.normal(3.5, 0.7, 200), 1, 5)
})
df['ventas'] = 2.5*df['ingresos'] + 1.1*df['gasto_marketing'] + np.random.normal(0, 8, 200)

# Gráfica
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Set 5: Correlaciones entre Variables del Negocio")
plt.show()
```

```python
# Datos
categorias6 = ['A', 'B', 'C', 'D']
valores6 = np.array([35, 25, 22, 18])

# Gráfica
plt.figure(figsize=(6, 6))
plt.pie(valores6, labels=categorias6, autopct='%1.1f%%', colors=sns.color_palette("pastel"))
plt.title("Set 6: Composición de Categorías")
plt.show()
```

```python
# Datos (Señal y FFT)
fs = 500
T = 2.0
t7 = np.linspace(0, T, int(fs*T), endpoint=False)
signal = 1.0*np.sin(2*np.pi*50*t7) + 0.6*np.sin(2*np.pi*120*t7) + 0.3*np.random.normal(0,1,len(t7))
Y = np.fft.rfft(signal)
freqs = np.fft.rfftfreq(len(t7), d=1/fs)
amplitud = np.abs(Y) / len(t7)

# Gráfica
plt.figure(figsize=(10, 5))
plt.plot(freqs, amplitud, color='blue')
plt.xlim(0, 150) # Limitamos para ver los picos de 50 y 120 Hz
plt.title("Set 7: Espectro de Potencia (Frecuencias)")
plt.xlabel("Frecuencia (Hz)")
plt.show()
```
